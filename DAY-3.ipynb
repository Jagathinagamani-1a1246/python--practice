```python
def alphabets(start, end):
    ch = ord(start)
    result = ""

    while ch <= ord(end):
        result = result + chr(ch) + " "
        ch += 1

    return result


print(alphabets('a', 'z'))
print(alphabets('A', 'Z'))
```

```python
def bill(units):
    if units <= 50:
        amount = units * 0.50
    elif units <= 150:
        amount = 50 * 0.50 + (units - 50) * 0.75
    elif units <= 250:
        amount = 50 * 0.50 + 100 * 0.75 + (units - 150) * 1.20
    else:
        amount = 50 * 0.50 + 100 * 0.75 + 100 * 1.20 + (units - 250) * 1.50
    
    surcharge = amount * 20 / 100
    return amount + surcharge

units = int(input("Enter units: "))
print("Electricity bill:", bill(units))
```

```python
def gross_salary(salary):
    if salary <= 10000:
        hra = salary * 20 / 100
        da = salary * 80 / 100
    elif salary <= 20000:
        hra = salary * 25 / 100
        da = salary * 90 / 100
    else:
        hra = salary * 30 / 100
        da = salary * 95 / 100

    return salary + hra + da

salary = float(input("Enter salary: "))
print("Gross salary:", gross_salary(salary))
```

```python
def marks(p,c,b,j,s):
    percentage = (p+c+b+j+s)/5
    if percentage >= 90:
        return "A"
    elif percentage >= 80:
        return "B"
    elif percentage >= 70:
        return "C"
    elif percentage >= 60:
        return "D"
    elif percentage >= 40:
        return "E"
    else:
        return "F"
Result= marks(90,80,70,60,50)
print("PERCENTAGE:",Result)
print("GRADE:",Result)
```
