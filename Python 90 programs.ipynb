```python
# 1. Maximum between two numbers

def maximum(a, b):
    if a > b:
        return a
    else:
        return b

a = int(input("Enter a: "))
b = int(input("Enter b: "))

print("Maximum:", maximum(a, b))
```

```python
# 2. Maximum between three numbers

def maximum(a, b, c):
    if a > b and a > c:
        return a
    elif b > c:
        return b
    else:
        return c

a = int(input("Enter a: "))
b = int(input("Enter b: "))
c = int(input("Enter c: "))

print("Maximum:", maximum(a, b, c))
```

```python
# 6. Leap Year

def leap_year(year):
    if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
        return "Leap Year"
    else:
        return "Not a Leap Year"

year = int(input("Enter year: "))

print(leap_year(year))
```

```python
# 12. Count total number of notes

def count_notes(amount):
    notes = [2000, 500, 200, 100, 50, 20, 10, 5, 2, 1]
    count = 0

    for note in notes:
        count = count + amount // note
        amount = amount % note

    return count

amount = int(input("Enter amount: "))

print("Total number of notes:", count_notes(amount))
```

```python
# 13. Calculate percentage and grade

def marks(p, c, b, m, comp):
    percentage = (p + c + b + m + comp) / 5

    if percentage >= 90:
        grade = "A"
    elif percentage >= 80:
        grade = "B"
    elif percentage >= 70:
        grade = "C"
    elif percentage >= 60:
        grade = "D"
    elif percentage >= 40:
        grade = "E"
    else:
        grade = "F"

    return percentage, grade

p = int(input("Enter Physics marks: "))
c = int(input("Enter Chemistry marks: "))
b = int(input("Enter Biology marks: "))
m = int(input("Enter Mathematics marks: "))
comp = int(input("Enter Computer marks: "))

percentage, grade = marks(p, c, b, m, comp)

print("Percentage:", percentage)
print("Grade:", grade)
```

```python
# 14. Calculate Gross Salary

def gross_salary(basic_salary):
    if basic_salary <= 10000:
        hra = basic_salary * 20 / 100
        da = basic_salary * 80 / 100
    elif basic_salary <= 20000:
        hra = basic_salary * 25 / 100
        da = basic_salary * 90 / 100
    else:
        hra = basic_salary * 30 / 100
        da = basic_salary * 95 / 100

    return basic_salary + hra + da

salary = float(input("Enter basic salary: "))

print("Gross salary:", gross_salary(salary))
```

```python
# 15. Electricity Bill

def electricity_bill(units):
    if units <= 50:
        amount = units * 0.50
    elif units <= 150:
        amount = 50 * 0.50 + (units - 50) * 0.75
    elif units <= 250:
        amount = 50 * 0.50 + 100 * 0.75 + (units - 150) * 1.20
    else:
        amount = (50 * 0.50 +
                  100 * 0.75 +
                  100 * 1.20 +
                  (units - 250) * 1.50)

    surcharge = amount * 20 / 100
    total = amount + surcharge

    return total

units = int(input("Enter electricity units: "))

print("Total electricity bill:", electricity_bill(units))
```

```python
# 16. Day of week using match-case

def week(day):
    match day:
        case 1:
            return "Monday"
        case 2:
            return "Tuesday"
        case 3:
            return "Wednesday"
        case 4:
            return "Thursday"
        case 5:
            return "Friday"
        case 6:
            return "Saturday"
        case 7:
            return "Sunday"
        case _:
            return "Invalid day"

day = int(input("Enter the number: "))

print(week(day))
```

```python
# 29. Palindrome number

def palindrome(n):
    original = n
    reverse = 0
    while n != 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10
    if original == reverse:
        return "Palindrome"
    else:
        return "Not Palindrome"
n = int(input("Enter number: "))
print(palindrome(n))
```

```python
# 33. Frequency of each digit

def digit_frequency(n):
    frequency = [0] * 10
    while n != 0:
        digit = n % 10
        frequency[digit] += 1
        n = n // 10
    result = ""
    for i in range(10):
        if frequency[i] > 0:
            result = result + str(i) + " : " + str(frequency[i]) + "\n"
    return result
n = int(input("Enter number: "))
print(digit_frequency(n))
```

```python
# 34. Number in words

def number_words(n):
    words = [
        "Zero", "One", "Two", "Three", "Four",
        "Five", "Six", "Seven", "Eight", "Nine"
    ]
    if n == 0:
        return "Zero"
    result = ""
    while n != 0:
        digit = n % 10
        result = words[digit] + " " + result
        n = n // 10
    return result
n = int(input("Enter number: "))
print(number_words(n))
```

```python
# 40. Prime numbers from 1 to n

def prime_numbers(n):
    result = ""
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            result = result + str(num) + " "
    return result
n = int(input("Enter n: "))
print("Prime numbers:", prime_numbers(n))
```

```python
# 41. Sum of prime numbers between 1 to n

def sum_primes(n):
    total = 0
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            total = total + num
    return total
n = int(input("Enter n: "))
print("Sum of prime numbers:", sum_primes(n))
```

```python
# 43. Armstrong number

def armstrong(n):
    original = n
    digits = len(str(n))
    total = 0
    while n != 0:
        digit = n % 10
        total = total + digit ** digits
        n = n // 10
    if total == original:
        return "Armstrong Number"
    else:
        return "Not an Armstrong Number"
n = int(input("Enter number: "))
print(armstrong(n))
```

```python
# 44. Armstrong numbers from 1 to n

def armstrong_numbers(n):
    result = ""
    for num in range(1, n + 1):
        original = num
        digits = len(str(num))
        total = 0
        temp = num
        while temp != 0:
            digit = temp % 10
            total = total + digit ** digits
            temp = temp // 10

        if total == original:
            result = result + str(num) + " "
    return result
n = int(input("Enter n: "))
print("Armstrong numbers:", armstrong_numbers(n))
```

```python
# 45. Perfect number

def perfect(n):
    total = 0
    for i in range(1, n):
        if n % i == 0:
            total = total + i
    if total == n:
        return "Perfect Number"
    else:
        return "Not a Perfect Number"
n = int(input("Enter number: "))
print(perfect(n))
```

```python
# 46. Perfect numbers from 1 to n

def perfect_numbers(n):
    result = ""
    for num in range(1, n + 1):
        total = 0
        for i in range(1, num):
            if num % i == 0:
                total = total + i
        if total == num:
            result = result + str(num) + " "
    return result
n = int(input("Enter n: "))
print("Perfect numbers:", perfect_numbers(n))
```

```python
# 47. Strong number

def strong(n):
    original = n
    total = 0
    while n != 0:
        digit = n % 10
        factorial = 1
        for i in range(1, digit + 1):
            factorial = factorial * i
        total = total + factorial
        n = n // 10
    if total == original:
        return "Strong Number"
    else:
        return "Not a Strong Number"
n = int(input("Enter number: "))
print(strong(n))
```

```python
# 48. Strong numbers from 1 to n

def strong_numbers(n):
    result = ""
    for num in range(1, n + 1):
        temp = num
        total = 0
        while temp != 0:
            digit = temp % 10
            factorial = 1
            for i in range(1, digit + 1):
                factorial = factorial * i
            total = total + factorial
            temp = temp // 10
        if total == num:
            result = result + str(num) + " "
    return result
n = int(input("Enter n: "))
print("Strong numbers:", strong_numbers(n))
```

```python
# 49. Fibonacci series

def fibonacci(n):
    a = 0
    b = 1
    result = ""
    for i in range(n):
        result = result + str(a) + " "
        c = a + b
        a = b
        b = c
    return result
n = int(input("Enter number of terms: "))
print("Fibonacci series:", fibonacci(n))
```

```python
# 50. Negative elements in an array

def negative_elements(arr):
    result = ""
    for i in arr:
        if i < 0:
            result = result + str(i) + " "
    return result
arr = list(map(int, input("Enter array elements: ").split()))
print("Negative elements:", negative_elements(arr))
```
