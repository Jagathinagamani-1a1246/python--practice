```python
# Using no arguments and no return value
def ncr(n,r):
    fn = 1
    fr = 1
    fnr = 1
    for i in range(1, n + 1):
        fn = fn * i
    for i in range(1, r + 1):
        fr = fr * i
    for i in range(1, n - r + 1):
        fnr = fnr * i
    result = fn // (fr * fnr)

    print("nCr =", result)
n = int(input("Enter n: "))
r = int(input("Enter r: "))
ncr(n,r)
```
